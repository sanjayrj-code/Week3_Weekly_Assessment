# Task 4 – Decision Framework and Proposal

## Objective
Evaluate five business problems using the Solution Decision Framework and select the most appropriate technology for each problem.

The evaluation considers:
- Input type
- Output type
- Latency requirements
- Cost
- Explainability
- Human review requirements

## Solution Decision Framework

| Business Problem | Input Type | Output Type | Latency | Cost | Explainability | Recommended Tool |
|---|---|---|---|---|---|---|
| GST calculation on invoices | Structured invoice data, taxable value, GST rate | Exact GST amount and total | Immediate | Low | High | Rule-based calculation system |
| Current leave balance | Employee ID and approved leave records | Exact remaining leave balance | Real-time | Low | High | SQL database query or HRMS API |
| Ranking resumes for a role | Resumes and job requirements | Ranked candidate list with reasons | Moderate | Moderate | Very high | Structured extraction and transparent scoring |
| Answering questions from a 60-page contract | Contract PDF and user questions | Answers with page references | Seconds | Moderate | High | Retrieval-Augmented Generation (RAG) |
| Forecasting next month's sales | Historical sales and business data | Sales forecast | Batch/daily | Low to moderate | Medium to high | Time-series forecasting model |

## Human Review and AI Limitations

1. **GST calculation:** Use rule-based calculations. A human should review unusual invoices, disputed tax classifications, and changes in GST rules. AI should not make authoritative tax decisions.

2. **Leave balance:** Retrieve balances directly from the HRMS database. HR staff should review disputed or exceptional leave records. AI should not invent or estimate leave balances.

3. **Resume ranking:** AI can assist with organizing and scoring resumes against job-related criteria. Human reviewers must check rankings and make all final hiring decisions. AI should not automatically reject candidates.

4. **Contract Q&A:** RAG can retrieve relevant contract clauses and generate answers with page references. A legal professional must review important legal interpretations and contractual decisions.

5. **Sales forecasting:** A forecasting model can estimate future sales. Managers should review forecasts against promotions, market changes, and unusual events. AI predictions should not be treated as guaranteed results.

# Technical Proposal: AI-Powered Contract Question Answering

## 1. Objective
Build a system that allows employees to ask questions about a 60-page contract and receive accurate answers supported by relevant page and section references.

## 2. Recommended Approach
Use Retrieval-Augmented Generation (RAG).

The system will:
- Extract text from the contract PDF.
- Divide the text into smaller sections (chunks).
- Convert chunks into embeddings.
- Store embeddings in a vector database.
- Retrieve relevant sections when a user asks a question.
- Use an LLM to generate an answer based only on the retrieved contract content.
- Include page references and state when the contract does not contain enough information.

## 3. Data Requirements
- The original 60-page contract in PDF format.
- Extracted text with page numbers.
- A set of sample questions and verified answers.
- An approved embedding model and LLM.
- Secure access permissions for contract users.

## 4. Evaluation Metrics and Targets
- Retrieval Recall@5: at least 90%.
- Answer correctness: at least 90% on a reviewed test set.
- Citation accuracy: at least 95%.
- The system must abstain when sufficient supporting information is unavailable.

These are proposed targets and must be validated through testing.

## 5. Estimated Build Time
- PDF extraction and preprocessing: 1 day.
- Chunking, embeddings, and vector database: 1–2 days.
- RAG and LLM integration: 2 days.
- Testing and evaluation: 1–2 days.

Estimated prototype time: 5–7 working days.

## 6. Key Trade-offs
- RAG provides contract-specific answers but requires document preprocessing.
- Embeddings improve semantic retrieval but add storage and processing costs.
- LLMs produce natural answers but can still make mistakes.
- Page citations improve traceability but must be checked for accuracy.
- Access controls and secure storage are necessary to protect confidential contracts.

## 7. Human Review and Security
Legal professionals must review important legal interpretations and decisions. The system should restrict access to authorized employees, protect contract data, maintain audit logs, and avoid using unsupported information.

## 8. Conclusion
A RAG-based contract question-answering system is recommended because it combines document retrieval with natural-language answers and traceable citations. It can reduce manual searching while keeping humans responsible for important legal decisions.

## Why Not Just Ask ChatGPT?

A general ChatGPT conversation may not have access to the company's confidential contract or the latest approved version.

A dedicated RAG system is better because:

1. **Document grounding:** It retrieves information directly from the company's contract.
2. **Citations:** It can provide page and section references for verification.
3. **Privacy:** Access controls can restrict contract information to authorized employees.
4. **Consistency:** It uses a controlled document and approved workflow.
5. **Evaluation:** Its answers can be tested against verified questions and answers.
6. **Human oversight:** It can send uncertain or important legal questions to a qualified reviewer.

ChatGPT can be useful for general explanations, but a dedicated RAG system provides better control and traceability for company-specific contract questions.
