# Task 3: Tokens, Cost, Temperature and Hallucination

```python
import tiktoken
import pandas as pd

print("Libraries imported successfully!")
```

```python
texts = [
    "Artificial Intelligence helps businesses automate repetitive tasks.",

    "Machine learning allows computers to learn patterns from historical data and make predictions.",

    "Natural language processing enables computers to understand, interpret and generate human language.",

    "Large Language Models use tokens to process text. They have context windows that limit how much information they can handle in a single request."
]

print("Total texts:", len(texts))
```

```python
# Load the GPT-4 compatible tokenizer
encoding = tiktoken.get_encoding("cl100k_base")

results = []

for i, text in enumerate(texts, start=1):

    words = len(text.split())

    tokens = len(encoding.encode(text))

    words_per_token = words / tokens if tokens > 0 else 0

    results.append({
        "Text": f"Text {i}",
        "Words": words,
        "Tokens": tokens,
        "Words per Token": round(words_per_token, 3)
    })

token_df = pd.DataFrame(results)

display(token_df)
```

```python
token_df.to_csv("token_analysis.csv", index=False)

print("Token analysis saved successfully!")
```

```python
# Monthly usage calculation

chats_per_day = 300
input_tokens_per_chat = 1200
output_tokens_per_chat = 300
days_per_month = 30

monthly_chats = chats_per_day * days_per_month

monthly_input_tokens = monthly_chats * input_tokens_per_chat
monthly_output_tokens = monthly_chats * output_tokens_per_chat

print("MONTHLY LLM USAGE")
print("-----------------")
print("Monthly chats:", monthly_chats)
print("Monthly input tokens:", monthly_input_tokens)
print("Monthly output tokens:", monthly_output_tokens)
print("Total monthly tokens:", monthly_input_tokens + monthly_output_tokens)
```

```python
# Enter the trainer's model pricing per 1 million tokens

input_price_per_million = 0.0
output_price_per_million = 0.0

monthly_input_cost = (
    monthly_input_tokens / 1_000_000
) * input_price_per_million

monthly_output_cost = (
    monthly_output_tokens / 1_000_000
) * output_price_per_million

monthly_total_cost = monthly_input_cost + monthly_output_cost

print("MONTHLY COST ESTIMATION")
print("-----------------------")
print("Input cost: $", round(monthly_input_cost, 2))
print("Output cost: $", round(monthly_output_cost, 2))
print("Total monthly cost: $", round(monthly_total_cost, 2))
```
